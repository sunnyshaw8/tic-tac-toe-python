```python
#Step 5: Write a function that uses the random module to randomly decide which player goes first. You may want to lookup random.randint() Return a string of which player went first.
```

```python
import random

def choose_first(a,b):
    result = random.randint(a,b)
    if result == a:
        return 'player 1 goes first'
    else:
        return 'player 2 goes first'
```

```python
#Step 6: Write a function that returns a boolean indicating whether a space on the board is freely available.
```

```python
def space_check(board, position):
    if board[position] == ' ':
        return True
    else:
        return False
```

```python
#Step 7: Write a function that checks if the board is full and returns a boolean value. True if full, False otherwise.
```

```python
def full_board_check(board):
    for i in board:
        if i == ' ':
            return False
        else:
            return True
```

```python
#step 8: Write a function that asks for a player's next position (as a number 1-9) and then uses the function from step 6 to check if it's a free position. If it is, then return the position for later use.
```

```python
def player_choice(board):
    pos = 0 
    
    while pos not in ([1,2,3,4,5,6,7,8,9]):
        pos = int(input("Please select a position from 0-9"))

        while space_check(board,pos) == False:
             print('sorry thats not a valid choice')
             pos = int(input("Please select a position from 1-9"))

    
    return pos
```

```python
def replay():

     choice= 'wrong'

     while choice not in ['yes','no']:

        choice = input("do you want to play again?reply in yes or no")

        if choice not in ['yes','no']:
            print("sorry thats not a valid choice")

     return choice
```

```python
print('welcome to tic tac toe game!')

# check if want to start
isready = ready()

if isready == 'yes':
    choice = player_input()

    # decide who goes first
    decision = choose_first(1, 2)
    print(decision)

    # display board
    game_on = True

    while game_on:
        board = [' ', ' ', ' ', ' ', ' ', ' ', ' ', ' ', ' ', ' ']
        playing = True
        player_mark = choice

        while playing:
            display_board(board)
            pos = player_choice(board)
            place_marker(board, player_mark, pos)
            display_board(board)

            if win_check(board, player_mark):
                playing = False
                print('congrats for winning')

                k = replay()

                if k == 'no':
                    print('thanks for playing')
                    game_on = False

            else:
                if full_board_check(board):
                    print('sorry the board is full match is drawn please restart the function')
                    break

                else:
                    if player_mark == 'x':
                        player_mark = 'o'
                    else:
                        player_mark = 'x'

else:
    print("hope you have a good day")
```
